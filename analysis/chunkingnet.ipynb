```python
def getchunk(inp_act,chunk,mnt1,mnt2):
    DInpMnt = []  #differnce between input and what is being maintained
    DInpChunk = [] #difference between input and chunk layer
    for i in range(len(inp_act)):
        Tmnt1 = mnt1[i]
        Tmnt2 = mnt2[i]
        Tinp = inp_act[i]
        Tchunk = chunk[i]

        DChunk = Tinp - Tchunk

        if Tmnt1!=-1 or Tmnt2!=-1:
            if Tmnt1!=-1 and Tmnt2!=-1:
                D1 = Tinp-Tmnt1
                D2 = Tinp-Tmnt2
                if abs(D1) < abs(D2):
                    DInpMnt.append(D1)
                else:
                    DInpMnt.append(D2)
                DInpChunk.append(Tinp-Tchunk)

            elif Tmnt1!=-1:
                D1 = Tinp-Tmnt1

                DInpMnt.append(D1)
                DInpChunk.append(Tinp-Tchunk)

            elif Tmnt2!=-1:
                D2 = Tinp-Tmnt2
                DInpMnt.append(D2)
                DInpChunk.append(Tinp-Tchunk)
    return DInpMnt,DInpChunk
```

Kind of confusing what is going on - why is it not symmetrical around 0? I think we have a combination of different things depending how both hid1 and hid2 are related to each other and the chunk - so lets break it down into different plots - 
1) inp - hid1 and inp - hid2 have the same sign
2) inp - hid1 and inp - hid2 have different signs
3) inp - hid1 and inp - hid2 is 0
4) inp - hid1 OR inp - hid2

```python
def getchunk_ideal(inp,chunk,hid1,hid2): 
    #since this is from the small chunk model, we do not have to take care of the fact there may or may not be 
    #be something for mnt1/mnt2 which is something we worry about for the main model. 
    
    #get all differences (not sure which will become useful)
    DInpChunk = inp-chunk
    DInpHid1 = inp-hid1
    DInpHid2 = inp-hid2
    DHid1Hid2 = hid1-hid2
    DHid1Chunk = hid1-chunk
    DHid2Chunk = hid2-chunk
    
    #lists of indices that fall under each of the following category
    samesign = []
    diffsign = []
    zeroand = []
    zeroor = []
    
    for i in range(len(DInpChunk)):
        if DInpHid1[i]*DInpHid2[i] > 0:
            samesign.append(i)
        elif DInpHid1[i]*DInpHid2[i] < 0:
            diffsign.append(i)
        elif DInpHid1[i] == 0 and DInpHid2[i] == 0:
            zeroand.append(i)
        elif DInpHid1[i] == 0 or DInpHid2[i] == 0 and DInpHid1[i]+DInpHid2[i] != 0:
            zeroor.append(i)
            
    return (samesign,diffsign,zeroand,zeroor,DInpChunk,DInpHid1,DInpHid2,DHid1Hid2,DHid1Chunk,DHid2Chunk)
            
def hid1hid2same(inp,chunk,hid1,hid2):
    DInpChunk = inp-chunk
    DInpHid1 = inp-hid1
    DInpHid2 = inp-hid2
    DHid1Hid2 = hid1-hid2
    DHid1Chunk = hid1-chunk
    DHid2Chunk = hid2-chunk
    
    DInpChunkS = []
    DInpHidS = []
    for i in range(len(DInpChunk)):
        if hid1[i] == hid2[i]: 
            DInpChunkS.append(DInpChunk[i])
            DInpHidS.append(DInpHid1[i])
    return DInpChunkS,DInpHidS



def chunkmin(inp,chunk,hid1,hid2):
    DInpHidMin = []  #differnce between input and what is being maintained
    DInpChunk = [] #difference between input and chunk layer
    for i in range(len(inp)):
        D1 = inp[i]-hid1[i]
        D2 = inp[i]-hid2[i]
        DInpChunk.append(inp[i]-chunk[i])
        DInpHidMin.append(min(D1,D2))

    return DInpHidMin,DInpChunk 
    
```

```python
#important definitions
def getchunk(inp_act,chunk,mnt1,mnt2):
    DInpMnt = []  #differnce between input and what is being maintained
    DInpChunk = [] #difference between input and chunk layer
    for i in range(len(inp_act)):
        Tmnt1 = mnt1[i]
        Tmnt2 = mnt2[i]
        Tinp = inp_act[i]
        Tchunk = chunk[i]

        DChunk = Tinp - Tchunk

        if Tmnt1!=-1 or Tmnt2!=-1:
            if Tmnt1!=-1 and Tmnt2!=-1:
                D1 = Tinp-Tmnt1
                D2 = Tinp-Tmnt2
                if abs(D1) < abs(D2):
                    DInpMnt.append(D1)
                else:
                    DInpMnt.append(D2)
                DInpChunk.append(Tinp-Tchunk)

            elif Tmnt1!=-1:
                D1 = Tinp-Tmnt1

                DInpMnt.append(D1)
                DInpChunk.append(Tinp-Tchunk)

            elif Tmnt2!=-1:
                D2 = Tinp-Tmnt2
                DInpMnt.append(D2)
                DInpChunk.append(Tinp-Tchunk)
    return DInpMnt,DInpChunk
def plot_chunking(files, savefile, plot_title = 'Chunking In Model', plot_xlabel = 'Input - Existing Stimuli', plot_ylabel = 'Input - Chunk'):
    for f in files:
        inp_act, chunk, stripe1, stripe2 = get_data(f)
        DInpMnt,DInpChunk = getchunk(inp_act,chunk,stripe1,stripe2)
        plt.scatter(DInpMnt,DInpChunk)
    plt.xlabel(plot_xlabel)
    plt.ylabel(plot_ylabel)
    plt.title(plot_title)
    plt.savefig(savefile)
    
    
def get_data(file):
    df = pd.read_csv(file, '\s+')
    df_include= [] #want all trials that are not recall
    for i in df['$TrialName']:
        if 'Recall' in i: 
            df_include.append('False')
        else:
            df_include.append('True')
    df.index=df_include
    df_norecall = df.loc['True']
    inp = [float(i.split('_')[1]) for i in df_norecall['$TrialName']]
    inp_act = df_norecall['#InDecode'].values
    chunk = df_norecall['#ChunkDecode'].values
    stripe1 = df_norecall['#stripe'].values
    stripe2 = df_norecall['#stripe\x01'].values
    return inp_act, chunk, stripe1, stripe2

    
```

```python
class Chunking():
    def __init__(self,file):
        df = pd.read_csv(file,'\s+')
        #df = pd.read_csv(file)
        self.df = df
        
    def chunk(self):
        df = self.df
        df_include= [] #want all trials that are not recall
        for i in df['$TrialName']:
            if 'Recall' in i:
                df_include.append('False')
            else:
                df_include.append('True')
        df.index=df_include
        df_norecall = df.loc['True']
        
        inp = [float(i.split('_')[1]) for i in df_norecall['$TrialName']]
        inp_act = df_norecall['#InDecode']
        chunk = df_norecall['#ChunkDecode']
        
        mnt1 = [float(i.split('_')[3]) for i in df_norecall['$TrialName']]
        mnt2 = [float(i.split('_')[5]) for i in df_norecall['$TrialName']]
        inp_act = inp_act.values
        chunk = chunk.values
        
        self.inp = inp
        self.mnt1 = mnt1
        self.mnt2 = mnt2
        self.inp_act = inp_act
        self.chunk = chunk
        
        num_trials = len(mnt1)
        DInpMnt = []  #differnce between input and what is being maintained
        DInpChunk = [] #difference between input and chunk layer
        for i in range(num_trials):
            Tmnt1 = mnt1[i]
            Tmnt2 = mnt2[i]
            Tinp = inp[i]
            Tchunk = chunk[i]

            DChunk = Tinp - Tchunk

            if Tmnt1!=-1 or Tmnt2!=-1:
                if Tmnt1!=-1 and Tmnt2!=-1:
                    D1 = Tinp-Tmnt1
                    D2 = Tinp-Tmnt2
                    if D1<D2 and abs(D1)>0.001:
                        DInpMnt.append(D1)
                        DInpChunk.append(Tinp-Tchunk)
                    if D2<D1 and abs(D2)>0.001:
                        DInpMnt.append(D2)
                        DInpChunk.append(Tinp-Tchunk)

                elif Tmnt1!=-1:
                    D1 = Tinp-Tmnt1
                    if abs(D1)>0.001:
                        DInpMnt.append(D1)
                        DInpChunk.append(Tinp-Tchunk)

                elif Tmnt2!=-1:
                    D2 = Tinp-Tmnt2
                    if abs(D2)>0.001:
                        DInpMnt.append(D2)
                        DInpChunk.append(Tinp-Tchunk)
        self.DInpMnt = DInpMnt
        self.DInpChunk = DInpChunk
        
    
        
    def plot(self):
        self.chunk()
        plt.scatter(self.DInpMnt,self.DInpChunk)
        plt.xlabel('Diff between Input and Maintained')
        plt.ylabel('Diff between Input and Chunk')
        plt.title('Chunk Layer')
```
