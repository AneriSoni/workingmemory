```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import os
sns.set()
from decodedoutputdefs import Precision #class with the compiled definitions
```

```python
base = 'Z:/leabra/examples/workingmemory/sir2_chunk/results/Ring/'
files = os.listdir(base)
lesion = Precision()
full_files = [base+f for f in files if 'LesionPFC_' in f] # need to use "LesionHidden_" "LesionPFC_" or "LesionPFCHidden
lesion.var_lesionprop(full_files,'PFC',["0","0.2","0.4","0.6","0.8","1"])
```

```python
error_dat = lesion.error_dat
bins = 45
i = 1
for key in error_dat.keys():
    plt.subplot(2,3,i)
    #plt.hist(diffrecall,range = (-0.01,0.01))
    #plt.hist(diffrecall,bins = 30, range = (-3,3))
    plt.hist(error_dat[key],bins = bins)
    plt.ylim([0,175])
    plt.title('LesionProp:{}'.format(key))
    i+=1
plt.tight_layout()
#plt.suptitle("PFC Lesion")
plt.show()
```
