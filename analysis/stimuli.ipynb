```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
#sns.set()
import scipy.stats as stats
import math
```

```python
mu = 3
variance = 1
sigma = math.sqrt(variance)
x = np.linspace(mu - 3*sigma, mu + 3*sigma, 100)
plt.plot(x, stats.norm.pdf(x, mu, sigma))
plt.xlim([0,20])
plt.xticks([])
plt.yticks([])
#plt.xlabel('Neuron')
#plt.ylabel('Activation')
#plt.title('Red')
plt.show()
```

```python
def count_to_hist(counts):
    num = len(counts)
    hist = np.linespace(0,10,10)
    data = []
    for h in range(num):
        h_data = np.repeat(h,counts[h])
        [data.append(i) for i in h_data]
    return data
        
    
```

```python
plt.hist(data)
plt.xlim([0,10])
plt.xticks([])
plt.yticks([])
plt.xlabel('Neuron')
plt.ylabel('Activation')
plt.title('Red')
```
