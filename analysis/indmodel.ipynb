```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import os
sns.set()
from decodedoutputdefs import Precision, training_curve_indv, training_curve_average, getfiles#class with the compiled definitions
```

```python
def select_best(base,num_best = 1):
    ABE = {}
    full_files = getfiles(base)
    for f in full_files:
        #import pdb; pdb.set_trace()
        model_num = f.split('model')[1].split('_')[0]
        df = pd.read_csv(f,'\s+')
        p = Precision()
        _,errrecall = p.get_diffs(df, nonresponse = 'guess')
        ABE[model_num] = np.mean(np.abs(errrecall))
    return ABE
```
