```python
x = np.linspace(-180,180, 181)
plt.plot(x,(np.exp((-np.abs(x))/20)))
plt.xlim([-180,180])
plt.ylim([-0.1,1.1])
plt.xlabel('Error in Color (Degrees)')
plt.ylabel('Reward')
plt.title('Continuous Exponential Reward')
```

```python
plt.plot(x,(np.exp((-np.abs(x))/10)))
plt.plot(x,(np.exp((-np.abs(x))/20)))
plt.plot(x,(np.exp((-np.abs(x))/30)))
plt.xlim([-180,180])
plt.ylim([-0.1,1.1])
plt.xlabel('Error in Color (Degrees)')
plt.ylabel('Reward')
plt.title('Continuous Exponential Reward')
plt.legend(['e^(-10x)', 'e^(-20x)', 'e^(-30x)'])
```
