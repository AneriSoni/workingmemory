In sir2_env, I added min dist and max dist that there should be between the two stimuli - and so lets check if that code is working. 
want to check the histogram of stim1 - stim2 

note: all models have stims from 0 to 360; so maximal possible difference is 180.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import os
sns.set()
from decodedoutputdefs import Precision #class with the compiled definitions
```

```python
def extract_mnt(df):
    
    mnt1 = []
    mnt2 = []
    for t in df['$TrialName']:
        mnt1t = (float(t.split('mnt1_')[1].split('_')[0]))
        mnt2t = (float(t.split('mnt2_')[1].split('_')[0]))
        if mnt1t==-1 or mnt2t==-1:
            pass
        else:
            mnt1.append(mnt1t)
            mnt2.append(mnt2t)
        
    return(mnt1,mnt2)

```

```python
base = 'C:/Users/Aneri/go/src/leabra/examples/sir_proj/sir2_chunk/checkstimrange/'
```

```python
#range 0 to 360
filename = '0to360.csv'
full_file = base+filename
df = pd.read_csv(full_file,sep = ',')
mnt1,mnt2 = extract_mnt(df)
stim = [0,360]
stim_diff = (stim[1]-stim[0])/2
plt.hist(np.mod(np.abs((np.array(mnt1)-np.array(mnt2))),stim_diff))
plt.xlabel('Differences')
plt.ylabel('Frequency')
plt.title(filename[:-4])
plt.show()
#this makes sense because the maximal possible differnece is 180 and so even though we set the range to 0 to 360, it doesn't make sense to have 
#a difference greater than 180.
```
